## 2. KPI validation

The following figures were carried forward from the Week 6 analysis and checked for internal consistency.

| KPI | Week 6 value | Week 7 validation | Status |
|---|---:|---:|---|
| Eligible appointments | 4,737 | 4,737 | Pass |
| Attended | 2,314 | 2,314 | Pass |
| No-show | 2,423 | 2,423 | Pass |
| Attendance rate | 48.85% | 48.85% | Pass |
| No-show rate | 51.15% | 51.15% | Pass |

The attendance and no-show counts add up to the total number of eligible appointments:

**2,314 + 2,423 = 4,737**

The two rates also add up to approximately 100%:

**48.85% + 51.15% = 100.00%**

This gives an internal check that the headline KPIs are consistent.


```python
# Simple KPI consistency check

eligible = 4737
attended = 2314
no_show = 2423

attendance_rate = attended / eligible * 100
no_show_rate = no_show / eligible * 100

print("Eligible appointments:", eligible)
print("Attended:", attended)
print("No-show:", no_show)
print("Attendance rate:", round(attendance_rate, 2), "%")
print("No-show rate:", round(no_show_rate, 2), "%")
print("Counts add up:", attended + no_show == eligible)
print("Rates add up:", round(attendance_rate + no_show_rate, 2))
```

### KPI test result

**Result: Pass**

The KPI calculations reproduce the Week 6 values. No correction was required for the headline attendance/no-show figures.

This means the main KPI values can continue to be used in the dashboard and project summary.


## 3. Validation of the main Week 6 finding

One of the main findings from Week 6 was that booking lead time showed a useful relationship with no-show behaviour.

The **31–60 day booking lead-time group** had the highest observed no-show rate in the Week 6 analysis.

A particularly important segment was:

> **Previous No-Show + 31–60 day booking lead time**

This segment had a **70.52% no-show rate across 977 appointments**.

For Week 7, I treated this as a finding to validate rather than automatically assuming that the recommendation was correct.


```python
# Week 6 priority-segment check
# These values are carried forward from the Week 6 analysis.

segment_appointments = 977
segment_no_show_rate = 70.52

print("Priority segment: Previous No-Show + 31–60 day booking lead time")
print("Appointments:", segment_appointments)
print("No-show rate:", segment_no_show_rate, "%")
```

### Segment validation result

**Result: Validated against the Week 6 analysis output.**

The priority segment remains the clearest high-risk segment identified in the Week 6 work.

However, this result should be treated as an observed pattern, not proof that booking lead time or previous no-show history causes a patient to miss an appointment.


## 4. Statistical validation carried forward from Week 6

Week 6 included statistical testing of booking lead-time distributions between no-show and attended appointments.

The Week 6 result provided evidence that the booking lead-time distributions differed between the two groups.

For Week 7, I kept this result as supporting evidence for the lead-time finding.

The important limitation is that a statistical association does not by itself establish causation. Therefore, the recommendation is framed around **risk-based monitoring and targeted reminders**, rather than claiming that changing booking lead time will definitely reduce no-shows.

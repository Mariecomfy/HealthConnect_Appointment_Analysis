## 7. Initial Data Quality Assessment

An initial assessment was conducted to evaluate the completeness, consistency, uniqueness, and general structure of the HealthConnect appointment dataset.

### Dataset completeness

The dataset contains 5,000 records across 18 variables. Most variables are complete. Missing values were identified in three variables:

- `reminder_channel`: 1,366 missing values
- `distance_to_clinic_km`: 90 missing values
- `waiting_time_minutes`: 60 missing values

The missing values in `reminder_channel` require further investigation because they may be related to whether a reminder was sent. Therefore, they should not automatically be treated as erroneous or removed.

### Duplicate and identifier assessment

There are no completely duplicated records in the dataset.

All 5,000 `appointment_id` values are unique, indicating that each record represents a distinct appointment.

There are 1,696 unique `patient_id` values among the 5,000 appointments. This indicates that patients may have multiple appointments within the dataset, which is relevant when considering previous appointment history and patient attendance patterns.

### Data types

The dataset contains appropriate data types for the initial analysis:

- Numerical variables include age, booking lead days, previous appointments, previous no-shows, distance to clinic, and waiting time.
- Categorical variables include gender, age group, appointment type, appointment day, reminder information, and appointment outcome.
- `booking_date` and `appointment_date` are stored as datetime variables.

### Appointment outcomes

The appointment outcome variable contains three categories:

- No-Show: 2,423
- Attended: 2,314
- Cancelled: 263

The presence of both cancelled and no-show appointments is important because these outcomes represent different appointment behaviours and should be considered separately during subsequent analysis.

### Initial assessment conclusion

The dataset appears suitable for further exploratory analysis. However, the missing values identified above and the treatment of cancelled appointments will need to be considered carefully during subsequent stages of the project.

## 10. Proposed Key Performance Indicators (KPIs)

The following KPIs are proposed for subsequent stages of the HealthConnect project:

| KPI | Definition | Linked Business Question | Relevance |
|---|---|---|---|
| **No-Show Rate** | Percentage of scheduled appointments that result in a no-show. | What proportion of scheduled appointments result in no-shows? | Provides a high-level measure of the clinic's appointment attendance problem and can be used to monitor improvement over time. |
| **Attendance Rate** | Percentage of scheduled appointments that are attended. | What proportion of scheduled appointments result in attendance? | Measures successful appointment attendance and provides a complementary view of the no-show rate. |
| **Reminder Effectiveness Rate** | Attendance rate among appointments where a reminder was sent, compared with appointments where no reminder was sent. | Is there a relationship between appointment reminders and patient attendance? | Helps assess whether the reminder process may contribute to improved appointment attendance. |
| **No-Show Rate by Appointment Type** | Percentage of appointments resulting in a no-show for each appointment type. | Are some appointment types associated with higher no-show rates? | Helps identify appointment categories that may require targeted engagement or intervention. |
| **Repeat No-Show Rate** | Proportion of appointments resulting in a no-show among patients with previous no-show history. | Does previous no-show behaviour appear to be associated with future no-shows? | Helps identify whether previous attendance behaviour may be useful for targeted patient engagement strategies. |

## 12. Assumptions, Limitations, Risks and Dependencies

### Assumptions

- The dataset contains fictional and anonymized appointment records.
- `appointment_outcome` is assumed to represent the final outcome of each scheduled appointment.
- Each `appointment_id` represents a unique appointment.
- Multiple appointments may belong to the same patient because `patient_id` is not unique.

### Limitations

- The dataset may not contain all factors that influence whether a patient attends an appointment.
- Missing values are present in reminder channel, distance to clinic, and waiting time.
- The dataset is fictional and may not fully represent real-world patient behaviour.
- Observed relationships between variables should not automatically be interpreted as causal relationships.

### Risks

- Incorrect treatment of cancelled appointments could distort no-show analysis.
- Missing data could affect comparisons between patient groups.
- Small groups within certain categories may produce less reliable comparisons.
- Analysis based on historical patterns may not necessarily predict future patient behaviour.

### Dependencies

- The HealthConnect Data Dictionary is required to correctly interpret the variables.
- Further analysis depends on validating the meaning and consistency of the available data.
- Subsequent recommendations will depend on the patterns identified during detailed analysis.
